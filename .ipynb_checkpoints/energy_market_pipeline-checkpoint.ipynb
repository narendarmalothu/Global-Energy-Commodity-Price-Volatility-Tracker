# Global Energy Market Data Extraction & EDA
This notebook extracts daily spot prices for major energy commodities using the EIA API, performs data cleaning, and conducts Exploratory Data Analysis (EDA) on price volatility before exporting the data for SQL ingestion.

```python
import requests
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configure visual style
sns.set_theme(style="darkgrid")
```

```python
# EIA API v2 Endpoint for Petroleum Spot Prices
API_KEY = "Api key used here(not to be disclosed)"
url = f"https://api.eia.gov/v2/petroleum/pri/spt/data/?api_key={API_KEY}&frequency=daily&data[0]=value&sort[0][column]=period&sort[0][direction]=desc"

try:
    response = requests.get(url)
    response.raise_for_status()
    data = response.json()
    print("Data successfully retrieved from EIA API.")
except requests.exceptions.RequestException as e:
    print(f"Error fetching data: {e}")
```

```python
# Convert JSON payload to Pandas DataFrame
df = pd.DataFrame(data['response']['data'])

# Data Cleaning & Feature Engineering
df['period'] = pd.to_datetime(df['period'])
df['value'] = pd.to_numeric(df['value'])

# Rename columns to standard database conventions
df.rename(columns={
    'period': 'price_date', 
    'value': 'spot_price', 
    'series-description': 'commodity_name'
}, inplace=True)

# Select relevant columns and drop nulls
df_clean = df[['price_date', 'commodity_name', 'spot_price']].copy()
df_clean.dropna(subset=['spot_price'], inplace=True)

# Simplify commodity names for easier visualization
def simplify_name(name):
    if 'Brent' in name: return 'Brent Crude'
    if 'WTI' in name: return 'WTI Crude'
    if 'Natural Gas' in name: return 'Natural Gas'
    return name

df_clean['commodity_name'] = df_clean['commodity_name'].apply(simplify_name)
print("Cleaned Data Sample:")
print(df_clean.head())
```

```python

# Exploratory Data Analysis (EDA) - Historical Price Trends
plt.figure(figsize=(14, 7))
sns.lineplot(data=df_clean, x='price_date', y='spot_price', hue='commodity_name')
plt.title("Historical Spot Prices of Major Energy Commodities", fontsize=16)
plt.xlabel("Date", fontsize=12)
plt.ylabel("Spot Price (USD)", fontsize=12)
plt.legend(title='Commodity')
plt.show()
```

```python

# Export cleaned dataset to CSV for MySQL ingestion
output_filename = "cleaned_energy_prices.csv"
df_clean.to_csv(output_filename, index=False)
print(f"\nDataset successfully exported to {output_filename}")
```
